## A damped spring
Type `\omega` and press TAB to get ω.

```fermium
k = 50 N/m
m = 0.5 kg
b = 0.2 kg/s
ω = √(k/m)
print ω in rad/s
```

```fermium
solve m x'' = -k x - b x'
  with x(0) = 10 cm, x'(0) = 0 m/s
  for t from 0 s to 5 s
print x(5 s)
plot x in cm vs t to "gallery/notebook_spring.png"
```
